```python
#overall churn rate
df["Churn"].value_counts()
```

```python
#churn by contract type has highest churn
df.groupby("Contract")["Churn"].value_counts(normalize=True)*100
```

```python
#churn by tenure
print(pd.crosstab(df["tenure"],df["Churn"],normalize="index")*100)
```

```python
#churn by monthly charges
print(df.groupby("Churn")["MonthlyCharges"].mean())
```

```python
#churn by payment method
df.groupby("PaymentMethod")["Churn"].value_counts(normalize=True)
```

```python
#churn by internet service
df.groupby("InternetService")["Churn"].value_counts(normalize=True)
```

```python
#churn by tech support
df.groupby("TechSupport").value_counts(normalize=True)
```

```python
#visualization
import matplotlib.pyplot as plt
```

```python
#overall churn
df["Churn"].value_counts().plot(kind="bar")
plt.title("customer Churn")
plt.xlabel("Churn")
plt.ylabel("number of customers")
plt.show()
```

```python
#churn by contract
pd.crosstab(df["Contract"],df["Churn"]).plot(kind="bar")
plt.title("Churn by Contract")
plt.xlabel("Contract")
plt.ylabel("number of customers")
plt.show()
```

```python
#Churn by payment method
pd.crosstab(df["PaymentMethod"],df["Churn"]).plot(kind="bar")
plt.title("Churn by PaymentMethod")
plt.xlabel("PaymentMethod")
plt.ylabel("number of customers")
plt.xticks(rotation=45)
plt.show()
```

```python
#Churn by monthly charges
df.boxplot(column="MonthlyCharges", by="Churn")
plt.title("monthly charges by churn")
plt.suptitle(" ")
plt.xlabel("Churn")
plt.ylabel("MonthlyCharges")
plt.show()
```

```python
#churn by tenure
df.boxplot(column="tenure",by="Churn")
plt.title("tenure by churn")
plt.suptitle(" ")
plt.xlabel("Churn")
plt.ylabel("tenure")
plt.show()
```
